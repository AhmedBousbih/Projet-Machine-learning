## 1. Chargement et vérification des données 

```python
df <- read.csv("healthcare_synthetic_data.csv", stringsAsFactors = FALSE)
```

Le jeu de données contient bien **15 000 patients** (nombre de ligne) et **19 variables** (nombre de colonne).  
La variable `Patient_ID` est un identifiant unique : elle sera **supprimée** pour l'analyse. On vérifie que la colonne Patient_ID est a valeur unique avec la commande

Les variables qualitatives sont déjà encodées sous forme d'entiers, il faudra les convertir en **facteurs** pour que R les traite correctement.

```python
str(df)
cat("\nValeurs manquantes par variable :\n")
print(colSums(is.na(df)))
df <- df %>% select(-Patient_ID)
```

Aucune valeur manquante n'est détectée dans le jeu de données.

### Analyse en composante princilpale

L'objectif de l'ACP est de résumer l'information contenue dans les variables quantitatives en un nombre réduit de composantes principales tout en conservant un maximum de variance.

Elle permet également de visualiser les corrélations entre variables et d'identifier d’éventuelles structures dans les données.

L'ACP est réalisée uniquement sur les variables quantitatives explicatives.



##### Sélection des variables quantitatives


```python
quant_cols <- c("Age", "Height_cm", "Weight_kg", "BMI",
                "Systolic_BP", "Diastolic_BP",
                "Cholesterol_Total", "Cholesterol_LDL", "Cholesterol_HDL",
                "Fasting_Blood_Sugar", "Stress_Level", "Sleep_Hours")

df_quant <- df[, quant_cols]
```

##### Réalisation de l’ACP

Les variables étant exprimées dans des unités différentes, on réalise une ACP réduite (standardisation des variables).

```python
acp <- PCA(df_quant,
           scale.unit = TRUE,
           graph = FALSE)
```

Ce graphique permet de déterminer combien de composantes principales conserver.




```python
fviz_eig(acp, addlabels = TRUE, ylim = c(0, 40))
```

Interprétation
La décroissance des valeurs propres montre que :

les premières composantes principales expliquent la majorité de la variance

on observe généralement un coude après les premières dimensions

En pratique, on retient souvent les 2 ou 3 premières composantes, qui résument l’essentiel de l’information du jeu de données.



##### Distribution des individus selon les composantes

On observe ici la dispersion des individus dans l’espace des composantes principales.

```python
g1 <- fviz_eig(acp, addlabels = TRUE, ylim = c(0,40))

g2 <- ggplot(melt(acp$ind$coord), aes(x = Var2, y = value)) +
  geom_boxplot() +
  xlab("Composantes principales") +
  ylab("Coordonnées des individus")

grid.arrange(g1, g2, ncol = 2)
```

Ce graphique permet de visualiser :

la dispersion des individus selon chaque composante

la présence éventuelle de valeurs atypiques (outliers)

Dans notre cas, la dispersion semble relativement homogène, ce qui suggère l’absence de valeurs aberrantes majeures dans les données.



##### Structure de corrélation des variables


```python
corrplot(acp$var$cor,
         is.corr = FALSE,
         method = "ellipse")
```

Ce graphique représente les corrélations entre les variables et les composantes principales.

On observe notamment :

une forte corrélation entre Weight_kg et BMI

une corrélation positive entre Cholesterol_Total et Cholesterol_LDL

une corrélation entre Systolic_BP et Diastolic_BP

Ces relations étaient déjà visibles dans la matrice de corrélation, ce qui confirme la cohérence de l’ACP.

```python
fviz_pca_var(acp)
```

Le cercle des corrélations permet d'analyser la relation entre les variables :

les variables proches l'une de l'autre sont positivement corrélées

les variables opposées sont corrélées négativement

les variables proches du cercle sont bien représentées par les deux premières composantes

On observe par exemple :

une proximité entre Weight_kg et BMI

une proximité entre Cholesterol_Total et Cholesterol_LDL

une relation entre Systolic_BP et Diastolic_BP

Ces résultats sont cohérents avec les relations médicales attendues.

##### Projection des individus

```python
fviz_pca_ind(acp,
             col.ind = "contrib",
             label = "none",
             gradient.cols = c("white", "#2E9FDF", "#FC4E07"))
```

Chaque point représente un patient projeté dans l’espace des deux premières composantes principales.

La couleur indique la contribution des individus à la formation des axes :

les individus les plus colorés contribuent le plus à la construction des axes

les individus proches du centre sont plus représentatifs de la moyenne

La distribution des points ne semble pas révéler de clusters clairement distincts, ce qui suggère une structure relativement continue des données.

##### Projection selon d’autres axes

```python
fviz_pca_var(acp, axes = c(1,3))

fviz_pca_ind(acp,
             col.ind = "contrib",
             label = "none",
             axes = c(1,3),
             gradient.cols = c("white", "#2E9FDF", "#FC4E07"))
```

L’observation d'autres axes permet d’explorer des relations supplémentaires entre variables qui ne sont pas visibles sur les deux premières composantes.

##### Visualisation selon la variable à prédire

On colore les individus selon Heart_Disease_Risk.



```python
fviz_pca_ind(acp,
             label = "none",
             habillage = df$Heart_Disease_Risk)
```

Cette représentation permet d’évaluer si les deux classes :

risque faible

risque élevé

sont séparables dans l’espace des composantes principales.

Si les deux groupes apparaissent relativement mélangés, cela suggère que :

la séparation entre les classes n’est pas parfaitement linéaire

des méthodes de classification plus complexes pourront être nécessaires.
